# Data Analysis and Visualization

## Calculate the monthly sales of the store and identify which month had the highest sales and which month had the lowest sales.

# 1. Monthly Sales Analysis

```python
sales_by_month = df.groupby("Order Month")["Sales"].sum().reset_index()
```

```python
sales_by_month
```

```python
plt.figure(figsize=(12, 6))
plt.plot(sales_by_month.index, sales_by_month.values,
         marker="o", color="tab:orange", linewidth=2)

plt.title("Monthly Sales Trend", fontsize=14)
plt.xlabel("Month", fontsize=14)
plt.ylabel("Sales", fontsize=14)  
plt.xticks(sales_by_month.index, rotation=45)
plt.grid(True, linestyle="--")
plt.tight_layout()
plt.show()
```

### Key Insights: Monthly Sales Trend

- **February** had the lowest sales, at about **59,751**.
- **November** had the highest sales, at about **352,461**.
- Sales stayed fairly steady in the middle of the year, then rose sharply in the last few months.
- This points to a **seasonal pattern**, so stock and marketing should be planned for the festive season.

## Analyze the sales and profit by customer segment

# 7. Sales and Profit by Customer Segment

```python
df.head()
```

```python
sales_profit_by_segment = df.groupby("Segment").agg({"Sales": "sum", "Profit": "sum"}).reset_index()
```

```python
sales_profit_by_segment
```

```python
# Convert the table to long format (needed for a grouped bar chart)
plot_df = sales_profit_by_segment.melt(
    id_vars="Segment",
    value_vars=["Sales", "Profit"],
    var_name="Metric",
    value_name="Amount"
)

plt.figure(figsize=(15, 7))
ax = sns.barplot(data=plot_df, x="Segment", y="Amount", hue="Metric", palette="Set2")

# Show values on top of the bars
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f", padding=3)

plt.title("Sales and Profit by Segment", fontsize=14)
plt.xlabel("Segment")
plt.ylabel("Amount")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## Key Insights: Sales and Profit by Segment

- **Consumer** is the biggest segment. It has the highest sales (**1,161,401**, about **50.6%** of total) and the highest profit (**134,119**, about **46.8%** of total).
- **Corporate** is second, with sales of **706,146** (30.7%) and profit of **91,979** (32.1%).
- **Home Office** is the smallest segment, with sales of **429,653** (18.7%) and profit of **60,299** (21.1%).
- **Profit margin** goes the other way. **Home Office** has the best margin at about **14.0%**, then **Corporate** at **13.0%**, and **Consumer** is lowest at **11.5%**.
- Consumer brings in half of all sales, but it earns less profit on each sale. Smaller segments are more efficient.

### Recommendations
- Keep Consumer as the main focus, since it gives the most revenue and profit.
- Review discounts and costs in Consumer to improve its margin. Even a 1% gain is large on 1.16M in sales.
- Grow Home Office and Corporate, because they have better margins.

```python
seg = df.groupby("Segment")[["Sales", "Profit"]].sum()
seg["Sales %"] = seg["Sales"] / seg["Sales"].sum() * 100
seg["Profit %"] = seg["Profit"] / seg["Profit"].sum() * 100
seg["Margin %"] = seg["Profit"] / seg["Sales"] * 100
print(seg.round(1))
```

## Analyze the sales to profit ratio

# 8. Sales to Profit ratio (Sale by Profit)

```python
sales_to_profit_by_segment = df.groupby("Segment").agg({"Sales": "sum", "Profit": "sum"}).reset_index()
sales_to_profit_by_segment["Sales_to_Profit_Ratio"] = sales_to_profit_by_segment["Sales"] / sales_to_profit_by_segment["Profit"]
print(sales_to_profit_by_segment[["Segment", "Sales_to_Profit_Ratio"]])
```

## > Note: Sales to Profit ratio = Sales ÷ Profit. A lower value means the segment earns more profit from each sale.
## Key Insights: Sales to Profit Ratio by Segment

- **Consumer** has the highest ratio at about **8.66**. It needs **8.66 in sales** to earn **1 in profit**.
- **Corporate** is next at about **7.68**.
- **Home Office** has the lowest ratio at about **7.13**, so it earns **1 in profit** from only **7.13 in sales**.
- A **lower ratio is better**. Home Office is the most efficient segment and Consumer is the least efficient.
- This matches the profit margin: Home Office has the best margin (**14.0%**) and Consumer has the lowest (**11.5%**).

### Recommendations
- Review discounts and costs in **Consumer** to bring its ratio down.
- Grow **Home Office** and **Corporate**, because they turn sales into profit more efficiently.

# 9. Final Summary: Key Insights and Recommendations

## Key Insights

### Overall Performance
- The store made about **2.30M in sales** and **286K in profit**, an overall profit margin of about **12.5%**.

### Seasonality
- Sales peak in **November (352K)** and are lowest in **February (60K)**.

### Product Performance
- **Technology** is the strongest category: **36.4% of sales** and **50.8% of profit**.

### Customer Segments
- **Consumer** is the biggest segment (**50.6% of sales**) but has the lowest margin (**11.5%**).

## Recommendations
1. **Fix Furniture:** review pricing, discounts and costs for Tables and Bookcases, because they cause most of the loss.
2. **Promote high-profit products:** use bundles and campaigns for Copiers, Accessories and Paper.
3. **Prepare for the peak season:** plan stock and marketing for September to December.
4. **Boost slow months:** use offers in January and February to reduce the gap.
5. **Improve Consumer margin:** reduce unnecessary discounts in the Consumer segment.
6. **Grow Home Office and Corporate:** they turn sales into profit more efficiently.

## Conclusion
Sales are healthy and spread evenly across categories, but **profit is not**. The biggest chance to improve is to **stop the losses in Tables and Bookcases** and **improve the Consumer margin**. Fixing these will raise profit without needing more sales.
